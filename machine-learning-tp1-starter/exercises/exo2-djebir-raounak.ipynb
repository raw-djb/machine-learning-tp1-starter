```python
from sklearn.datasets import fetch_20newsgroups
newsgroups = fetch_20newsgroups(subset="train")
```

```python
print("number of documents:", len(newsgroups.data))
print("number of labels:", len(newsgroups.target))
```

```python
print("first document:")
print(newsgroups.data[0])
print("target label:")
print(newsgroups.target[0])
```

```python
print(newsgroups.target_names)
print("first document label:", newsgroups.target[0], " -- ", newsgroups.target_names[newsgroups.target[0]])
```

```python
import numpy as np
from collections import Counter
lengths = [len(doc.split()) for doc in newsgroups.data]
print("mean length (words):", np.mean(lengths), "max:", np.max(lengths))
```

```python
words = Counter(w.lower() for doc in newsgroups.data[:1000] for w in doc.split())
print(words.most_common(10))
```
